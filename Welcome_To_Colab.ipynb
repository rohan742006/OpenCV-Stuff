```
import cv2
from google.colab.patches import cv2_imshow
image=cv2.imread('id.jpg')
gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```
cv2_imshow(image)
```

```
small=cv2.resize(image,(400,300))
cv2_imshow(small)
```

```
image = cv2.imread('id.jpg')
cv2.putText(image,'Rohan',(500,500),cv2.FONT_HERSHEY_SIMPLEX,1,(255,0,0),2)
cv2_imshow(image)
```

```
line=cv2.line(image,(0,0),(250,250),(0,0,512),10)
cv2_imshow(line)
```

```
circ=cv2.circle(image,(630,200),80,(0,0,255),5)
cv2_imshow(circ)
```

```
rect=cv2.rectangle(image,(1000,50),(200,600),(0,0,255),5)
cv2_imshow(rect)
```
